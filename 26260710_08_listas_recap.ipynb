```python
### Ejercicio 1 

numeros = [4, 7, 2, 4, 9, 7, 1, 2, 10, 4]

nueva_lista = []

for numero in numeros:
    if numero not in nueva_lista:
        nueva_lista.append(numero)

print(nueva_lista)
```

```python
### Ejericio 2

numeros = [12, 7, 9, 24, 31, 18, 5, 42, 13, 8]

pares = []
impares = []

for numero in numeros:
    if numero % 2 == 0:
        pares.append(numero)
    else:
        impares.append(numero)

print("Pares:", pares)
print("Impares:", impares)
```

```python
### Ejercicio 4

palabras = ["python", "sol", "programacion", "lista", "computadora", "if", "variable"]

contador = 0

for palabra in palabras:
    if len(palabra) > 5:
        contador = contador + 1

print("Cantidad:", contador)
```

```python
### Ejercicio 5

notas = [14, 18, 11, 16, 9, 13, 20]

suma_total = 0
total_notas = len(notas)

for nota in notas:
    suma_total = suma_total + nota

promedio = suma_total / total_notas

print(f"El promedio de notas es:{promedio}")

if promedio >= 12:
    print("Aprobó")
else:
    print("Desaprobó")
```

```python
### Ejercicio 9

lista1 = [4, 8, 12, 16, 20]
lista2 = [8, 10, 12, 18, 20] 

unicos = []

for numero in lista1:
    if numero not in lista2:
        unicos.append(numero)

for numero in lista2:
    if numero not in lista1:
        unicos.append(numero)

print(unicos)
```

```python
### Ejercicio 11

palabras = ["casa", "elefante", "sol", "computadora", "mesa", "programacion"]

longitudes_palabras = [
    len(palabra)
    for palabra in palabras

]
print(f"Longitudes palabras: {longitudes_palabras}")

lista_listas = []
for item in zip(palabras, longitudes_palabras):
    lista_listas.append(item)

sorted(lista_listas, key=lambda x:[1])
```

```python
### Ejercicio 13

nombres = ["Ana", "Luis", "Ana", "Carlos", "Luis", "Ana", "María", "Carlos"]

contados = []

for nombre in nombres:
    if nombre not in contados:
        cantidad = nombres.count(nombre)
        print(nombre, "aparece", cantidad, "veces")
        contados.append(nombre)
```
