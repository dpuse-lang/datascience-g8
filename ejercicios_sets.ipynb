```python
### Ejercicio 1

estudiantes_python = {
"Ana", "Luis", "María", "Carlos", "Sofía", "Diego"
}
estudiantes_javascript = {
"María", "Carlos", "Pedro", "Lucía", "Diego", "Valeria"
}

print("Estudiantes solo en Python", estudiantes_python - estudiantes_javascript)
print("Estudiantes solo en Javascript", estudiantes_javascript - estudiantes_python)

print("Todos los estudiantes", estudiantes_python | estudiantes_javascript)
```

```python
### Ejercicio 2 

almacen_arequipa = {
"laptop", "teclado", "mouse", "monitor", "impresora", "audifonos"
}
almacen_lima = {
"laptop", "mouse", "monitor", "tablet", "webcam", "parlantes"
}

print("productos en ambos almacenes", almacen_arequipa & almacen_lima)
print("productos solo arequipa", almacen_arequipa - almacen_lima)
print("productos solo lima", almacen_lima - almacen_arequipa)
print("todos los procuctos unicos", almacen_arequipa | almacen_lima)
print("¿mismos inventario?", almacen_arequipa == almacen_lima)
```

```python
### Ejercicio 3: 

permisos_usuario = {
"leer", "editar", "comentar", "descargar"
}
permisos_requeridos = {
"leer", "editar", "eliminar", "publicar"
}

print("permisos que le faltan", permisos_requeridos - permisos_usuario)
print("permisos adicionales", permisos_usuario - permisos_requeridos)
print("permisos totales", permisos_requeridos | permisos_usuario)
print("permisos diferentes:", permisos_requeridos ^ permisos_usuario)
print("total de permisos", len("permisos diferentes"))
```
