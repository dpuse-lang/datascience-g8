```python
from matplotlib import pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 45]


plt.plot(x, y, marker='o', linestyle='-', color='b')
plt.xlabel('Periodo')
plt.ylabel('Valor')
plt.title('Ejemplo de gráfico de líneas')
plt.show()
```

```python
categorias = ['A', 'B', 'C', 'D']
valores = [40, 70, 30, 90]

plt.bar(categorias, valores, color='#B628FA')
plt.xlabel('Categoria')
plt.ylabel('Frecuencia')
plt.title('Gráfico de barras verticales')
plt.show()
```

```python
plt.barh(categorias, valores, color='green')
plt.xlabel('Frecuencia')
plt.ylabel('Categoria')
plt.title('Gráfico de barras horizontal')
plt.show()
```
